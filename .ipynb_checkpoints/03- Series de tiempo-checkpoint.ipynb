```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
```

```python
dta= sm.datasets.co2.load_pandas().data
```

```python
dta.plot();
plt.title("Niveles de C02")
plt.ylabel("Partes por millon")
plt.xlabel("Año")
plt.show();
```

```python
import yfinance as yf
import numpy as np
```

```python
ticker_google = "GOOGL"
start_date= "2020-01-01"
end_date = "2025-12-31"
dataframegoogle2025 = yf.download(ticker_google, start=start_date, end=end_date)
```

```python
dataframegoogle2025
```

```python
dataframegoogle2025["Close"].plot()
plt.title("Eje normal")
plt.show();
        
```

```python
dataframegoogle2025.info()
```

```python
dataframegoogle2025["LogClose"] = np.log(dataframegoogle2025["Close"])
```

```python
dataframegoogle2025
```

```python
dataframegoogle2025["LogClose"].plot()
plt.title("Eje logarítmico")
plt.show();
```
