## Clasificadores

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn
```

```python
from sklearn.metrics import roc_curve, auc
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
```

```python
# Formato a usar: nombre del clasificador -- (formato de línea, nombre de clasificación)
Class_map = {"LogisticRegression": ("-", LogisticRegression()), 
             "NaiveBayes": ("--", GaussianNB()),
             "Decision tree": (".-", DecisionTreeClassifier(max_depth=5)),
             "Random forest": (":", RandomForestClassifier(max_depth=5, n_estimators=10, max_features=1))
            }
df = sns.load_dataset("iris")
df
```

```python
# Se realizan las condiciones para ver si son virginicas.
X, Y = df[df.columns[:3]], (df["species"]=="virginica")
Y
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size= 0.8)
```

```python
for nombre, (line_fmt, model) in Class_map.items():
    model.fit(X_train, Y_train)
    preds = model.predict_proba(X_test)
    prediccion = pd.Series(preds[:,1])
    fpr, tpr, umbral = roc_curve(Y_test, prediccion)
    auc_score = auc(fpr, tpr)
    etiqueta = "%s: auc=%f" % (nombre, auc_score)
    plt.plot(fpr, tpr, line_fmt, linewidth=5, label=etiqueta)
plt.legend(loc="lower right")    
plt.title("Comparación de clasificadores")
plt.plot([0,1], [0,1], "k--")
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel("Indices de falsos positivos")
plt.ylabel("Indices de verdaderos positivos")
plt.show()
```
